```python
import numpy as np
import os
from scipy.io import savemat,loadmat,whosmat
import numpy.ma as ma
import xarray as xr
import matplotlib.pyplot as plt
import pandas as pd
```

```python
piC_files = sorted([f for f in os.listdir('../processed_netcdf/piControl/CMIP6_sia_sh') if '.nc' in f])
unique_id = [f.split('_')[2]+'_'+f.split('_')[3] for f in piC_files]
have_pic = pd.DataFrame({'name' : unique_id, 'have_SH_piC' : ['yes']*len(unique_id)})
have_pic = have_pic.set_index('name')
print(have_pic)
```

```python
siconc_files = sorted([f for f in os.listdir('../processed_netcdf/CMIP6_sic_1979-2014_clim') if '.nc' in f])
unique_id = [f.split('_')[2]+'_'+f.split('_')[4] for f in siconc_files]
have_sic = pd.DataFrame({'name' : unique_id, 'have_SH_sic' : ['yes']*len(unique_id)})
have_sic = have_sic.set_index('name')
print(have_sic)
```

```python
siv_files = sorted([f for f in os.listdir('../processed_netcdf/CMIP6_sivol_1979-2014_clim') if '.nc' in f])
unique_id = [f.split('_')[2]+'_'+f.split('_')[4] for f in siv_files]
have_siv = pd.DataFrame({'name' : unique_id, 'have_SH_siv' : ['yes']*len(unique_id)})
have_siv = have_siv.set_index('name')
print(have_siv)
```

```python
model_ds = xr.open_dataset('../processed_netcdf/processed_CMIP6_sia_historical_r1_updated.nc')
have_sia = pd.DataFrame({'name' : model_ds.name.values, 'have_SH_sia' : ['yes']*len(model_ds.name.values)})
have_sia = have_sia.set_index('name')
print(have_sia)
```

```python
alldf = have_sia.join(have_sic,how='left') # this was outer
alldf = alldf.join(have_pic,how='left')
alldf = alldf.join(have_siv,how='left')
alldf = alldf.where(alldf=='yes','no')
```

```python
ensemble_ds = xr.open_dataset('../processed_netcdf/processed_CMIP6_sia_historical_allr_updated.nc')
enames = ensemble_ds.name.values

emod = [f.split('_')[0]+'_' for f in enames]
umod = list(set(emod))
nens = []
for u in umod:
    nens.append(len([f for f in emod if u in f]))
fullnames = alldf.index.values

fullbyu = []
for u in umod:
    fullbyu.append([f for f in fullnames if u in f][0])

no_ens = pd.DataFrame({'name' : fullbyu, 'no_ens_historical' : nens})
no_ens = no_ens.set_index('name')
newdf = alldf.join(no_ens,how='left')
print(newdf)
newdf['no_ens_historical'] = newdf['no_ens_historical'].fillna(0)
newdf['no_ens_historical'] = newdf['no_ens_historical'].astype(int)
```

```python
ensemble_ds = xr.open_dataset('../processed_netcdf/processed_CMIP6_sia_ssp245_allr_updated.nc')
enames = ensemble_ds.name.values

emod = [f.split('_')[0]+'_' for f in enames]
umod = list(set(emod))
nens = []
for u in umod:
    nens.append(len([f for f in emod if u in f]))
fullnames = alldf.index.values

fullbyu = []
for u in umod:
    fullbyu.append([f for f in fullnames if u in f][0])

no_ens_ssp245 = pd.DataFrame({'name' : fullbyu, 'no_ens_ssp245' : nens})
no_ens_ssp245 = no_ens_ssp245.set_index('name')
newerdf = newdf.join(no_ens_ssp245,how='left')
newerdf['no_ens_ssp245'] = newerdf['no_ens_ssp245'].fillna(0)
newerdf['no_ens_ssp245'] = newerdf['no_ens_ssp245'].astype(int)
newerdf.index = [f.split('_')[0] for f in newerdf.index.values]
print(len(newerdf))
newerdf
```

```python
first_member_names = xr.open_dataset('../processed_netcdf/processed_CMIP6_sia_historical_r1_updated.nc').name.values
firstname = [f.split('_')[0] for f in first_member_names]
firstens = [f.split('_')[1] for f in first_member_names]

firstds = pd.DataFrame({'name' : firstname, '1st member (hist)' : firstens})

firstds = firstds.set_index('name')

#firstmem['first member'] = [f.split('_')[1] for f in newerdf.index.values]
newerdf = newerdf.join(firstds,how='left')
newerdf
```
